# Lambda an Anonymous function:
It's such small function having lambda as keyword which can take any nums of args but is limited to single expression.
The expression's result is auto. returend w/o needing to return statement.
syntax: lambda *args:expression (function or operation to apply)

```python
f = lambda n:n**(1/2)
print(f(4))
```

```python
# Passing the lambda function as arg inside another function
def func(fx, n1,n2,n3,n4):
    return fx(n1,n2,n3,n4)

a,b,c,d = int(input("Enter num:")),int(input("Enter num:")),int(input("Enter num:")),int(input("Enter num:"))
print(f"Max is {func( (lambda a,b,c,d: max(a,b,c,d)) ,a,b,c,d)}")
```

```python
# Applying the customize behaviour in the expression section of lambda function
# This lambda Will add only the odd numbers in the list and return theirs sum
def My_func(my_list):
    total =0
    for n in my_list:
        if(n%2==1):
            total +=n
    return total
    
ls = [4,3,1,7,9,5,11]
F  = lambda ls:My_func(ls)
print(F(ls))
```

```python
# Returning lambda function from another function
def Fx(n):
    return lambda x:x*n
    
n = int(input("Enter the value of multiplicant: "))    
f = Fx(n) # after fx is done this will be look like f = lambda x:x*n ( n will be replaced by the number which was given by user)
x = int(input('Enter the value of multiplier: '))
print(f"The resultant is {f(x)}")
```

```python
# Applying the lambda function inside another function
def Func(n,x):
    f = lambda x:x**n
    return f(x)

print(f"Ans is {Func(2,4)}")
```
